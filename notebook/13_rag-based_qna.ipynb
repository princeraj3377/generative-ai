```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_google_genai import ChatGoogleGenerativeAI, GoogleGenerativeAIEmbeddings
from langchain_chroma import Chroma
```

```python
loader = PyPDFLoader("../data/data_science_syllabus.pdf")
doc= loader.load()
len(doc)
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 1000, chunk_overlap = 200)
splitter_data= splitter.split_documents(doc)
len(splitter_data)
```

```python
# text-embedding-004 was retired; use the supported text embedding model.
embeddings = GoogleGenerativeAIEmbeddings(model="models/gemini-embedding-001")
```

```python
vector_store = Chroma.from_documents(
    documents=splitter_data,
    embedding=embeddings
)
```

```python
query = "machine learning and data science"
data = vector_store.similarity_search(query=query)
```

```python
len(data)
```

```python
data[0]
```

```python
context = ""
for doc in data:
    context += doc.page_content +"/n"
```

```python
print(context)
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash")
```

```python
res = llm.invoke(f"""can tou provid me the answer based on 
                 provide d context :{context}for my question {query}""")
```

```python
print(res.content)
```
