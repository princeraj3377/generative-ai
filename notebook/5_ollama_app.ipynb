```python
from pydantic import  BaseModel,Field
```

```python
class respose_structure(BaseModel):
    name: str=Field(...,description="name of the person")
    email: str=Field(...,description="email of the person")
    age: int=Field(...,description="age of the person")

    
```

```python
structure = llm.with_structured_output(
    respose_structure,
    method="json_mode"
)
res = structure.invoke(
    f"Extract the person's details from this text. Return ONLY valid JSON with exactly these keys: name, email, age. No Markdown or extra text. Text: {text}"
)
print(res.model_dump())
```
