```python
from dotenv import load_dotenv
load_dotenv()  # Load environment variables from .env file
```

```python
from langchain_core.tools import tool
```

```python
@tool
def get_sum(a : int , b :int ):
    """This tool takes two integers as input and returns their sum.


    Args:
        a (int): The first integer.
        b (int): The second integer.



    """
    return a+b
```

```python
res=get_sum.invoke({"a": 5, "b": 10}    )
res
```

```python
from langchain_ollama import ChatOllama
from langchain.agents import create_agent
```

```python
llm = ChatOllama(model="gemma4:31b-cloud")
```

```python
agent = create_agent(
    model = llm,
    tools=[get_sum],
    system_prompt="You are a helpful assistant that can perform mathematical operations. Use the provided tools to answer questions."
)
```

```python
response = agent.invoke({"messages":[{"role":"user","content":"What is the sum of 15 and 25?"}]})
```

```python
for mchunk in response['messages']:
    print(mchunk)
    print("/n")
```
