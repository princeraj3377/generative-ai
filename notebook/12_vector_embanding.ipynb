```python
from dotenv import load_dotenv
load_dotenv()
```

```python
document =[
    "Machine learning is a branch of artificial intelligence ",
    "that trains computers to learn from data and make predictions ",
    "without being explicitly programmed for every single task."
]
```

```python
from langchain_ollama import OllamaEmbeddings
```

```python
embading = OllamaEmbeddings(model= "qwen3-embedding")
```

```python
vector = embading.embed_documents(document)
```

```python
vector
```

### vector store 

```python
from langchain_chroma import Chroma
```

```python
vextor_store = Chroma.from_texts(texts= document,embedding=embading,persist_directory="./vector_db")
```

```python
query = "machine learning"
result = vextor_store.similarity_search(query = query)
```
