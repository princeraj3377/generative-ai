```python
### firld validation 
from pydantic import Field

class product(BaseModel):
    title :str = Field(max_length=10)
    description :str= Field(min_length=20)
    price : int =Field(gt=0,le=100)
```

```python
pr = product(title="sasumshs", description="thbajk,nhfnsfahbfmhbsfjhgsbafjhsa", price=34)
pr
```

```python
## now came on multiple class validation 
class Order(BaseModel):
    id:str =Field(min_length=6,max_length=12)
    pri:product
```

```python
order = Order(id="qwddsdfdsf",pri=pr )
```

```python
order
```

```python
print(order.pri.title)
print(order.pri.price)
```

```python
from typing import List
class Card(BaseModel):
    id:str=Field(min_length=6,max_length=12)
    items: List[product]#[product1,product2.....] here comes typing laibary
    comment :str = Field(
        default="No comment "
    )
```

```python
card = Card(id="vskjnfhjds",items=[pr])
```

```python
card
```

```python
from typing import Annotated
##annotated provide him to create a common validation 
commonValidator = Annotated[List[product], Field(default=[])]
```

```python
class Card(BaseModel):
    id:str=Field(min_length=6,max_length=12)
    items: commonValidator
    comment :str = Field(
        default="No comment "
    )
```

```python
class Order(BaseModel):
    products:commonValidator
```

```python

ord = Order(products=[pr])
ord
```
