```python
from dotenv import load_dotenv
load_dotenv()  # Load environment variables from .env file
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain.agents import create_agent
from langchain_ollama import ChatOllama
from langgraph.checkpoint.memory import MemorySaver

```

```python
search = GoogleSerperAPIWrapper()
```

```python
llm= ChatOllama(model="gemma4:31b-cloud")
```

```python
agent = create_agent(model = llm , tools = [search.run],system_prompt = "you are a helpful assistant that can search the web using Google Search API. You will be given a query and you should return the most relevant information from the search results. If you cannot find any relevant information, you should say 'I could not find any relevant information on this topic.",checkpointer=MemorySaver())
```

```python
question = "how many follwer are in the instagram account os @theprinceraj.in"
respone = agent.invoke({"messages":[{"role":"user","content":question}]},
                       {"configurable":{"thread_id":"prince"}}
                       )
```
