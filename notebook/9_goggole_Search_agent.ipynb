```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_ollama import ChatOllama
from langchain.agents import create_agent
```

```python
search = GoogleSerperAPIWrapper()
```

```python
llm= ChatOllama(model="gemma4:31b-cloud")
```

```python
agent = create_agent(model =llm, tools=[search.run], system_prompt="You are a helpful assistant that can answer questions and perform searches using the Google Search API. Use the search tool when necessary to find information.")
```

```python
question = "how many followers are in the instagram account of @theprinceraj.in"
response = agent.invoke({"messages": [{"role": "user", "content": question}]})
```

```python
print(response['messages'][-1].content)
```
